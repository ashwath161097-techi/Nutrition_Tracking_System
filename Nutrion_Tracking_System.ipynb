**NUTRITION TRACKING SYSTEM**

**PROJECT OBJECTIVE**


The Nutrition Tracker on service delivery and beneficiary coverage under the Integrated Child Development Services. It records the number of Anganwadi centres, workforce availability, beneficiary groups, nutrition services, and infrastructure conditions across states and districts in India. The data is organised by reporting date and reflects programme operations and beneficiary coverage under the national nutrition monitoring system.

**PROBLEM STATMENT**

The Ministry of Women and Child Development needs to optimize the delivery of nutritional services. The primary goals are:

To identify districts with high rates of malnutrition (stunting, wasting, and underweight) that require urgent intervention.

To evaluate the infrastructure gap (e.g., lack of drinking water or functional toilets) in AWCs and its impact on service delivery.

To assess the reach of Hot Cooked Meals (HCM) and Take-Home Rations (THR) among eligible beneficiaries.

**EXPECTED OUTCOME**

A data-driven prioritization matrix that ranks districts based on nutritional risk and infrastructure needs, helping policymakers allocate resources more effectively.

```python
# Rename complex columns for better legibility
rename_dict = {
    'awc_infra_drink_water_src': 'awc_with_drinking_water',
    'awc_infra_func_toilets': 'awc_with_functional_toilets',
    'awc_infra_own_building': 'awc_owned_building',
    'severely_moderately_underweight': 'pct_underweight_children',
    'severely_or_moderately_stunted': 'pct_stunted_children',
    'wasting_sam_or_mam': 'pct_wasting_children'
}
df.rename(columns=rename_dict, inplace=True)
```

```python
# Convert date string to datetime
df['date'] = pd.to_datetime(df['date'])
```

```python
# Convert numeric codes to strings
df['state_code'] = df['state_code'].astype(str)
df['district_code'] = df['district_code'].astype(int).astype(str)
```

**11 Data Transformation & Wrangling**

```python
# Basic Utility Coverage Percentages (safely handling division by zero)
df['infra_water_coverage_pct'] = np.where(df['anganwadi_centers'] > 0,
                                          (df['awc_with_drinking_water'] / df['anganwadi_centers']) * 100, 0)

df['infra_toilet_coverage_pct'] = np.where(df['anganwadi_centers'] > 0,
                                           (df['awc_with_functional_toilets'] / df['anganwadi_centers']) * 100, 0)
```

```python
# 2. Extract Absolute Counts of Malnourished Children from Percentages
# (Crucial for logistical supply chain planning for food and health supplies)
df['est_stunted_children_count'] = (df['pct_stunted_children'] / 100) * df['children_measured_0_6_years']
df['est_underweight_children_count'] = (df['pct_underweight_children'] / 100) * df['children_measured_0_6_years']
```

**STATISTICAL ANALYSIS**

Selecting Numarical Columns

```python
# Select representative numerical columns
columns_to_analyze = [
    'anganwadi_centers',
    'pct_underweight_children',
    'pct_stunted_children',
    'pct_wasting_children',
    'eligible_beneficiaries'
]
# Clean missing values for the selected analysis columns
df_clean = df[columns_to_analyze].dropna()
```

Measuring  Central Tendency

```python
print("==================================================")
print("1. MEASURES OF CENTRAL TENDENCY")
print("==================================================")

for col in columns_to_analyze:
    print(f"\n--- Feature: {col} ---")

    # 1. Mean calculation
    mean_val = df_clean[col].mean()
    print(f"Mean (Average)      : {mean_val:.2f}")

    # 2. Median calculation
    median_val = df_clean[col].median()
    print(f"Median (Middle)     : {median_val:.2f}")

    # 3. Mode calculation (takes the first mode if multiple exist)
    mode_val = df_clean[col].mode().iloc[0]
    print(f"Mode (Most Frequent): {mode_val:.2f}")
```

Variance and Standard Deviation

```python
print("\n==================================================")
print("2. VARIANCE AND STANDARD DEVIATION")
print("==================================================")

for col in columns_to_analyze:
    print(f"\n--- Feature: {col} ---")

    # 1. Variance calculation (Sample variance by default: ddof=1)
    variance_val = df_clean[col].var()
    print(f"Variance           : {variance_val:.2f}")

    # 2. Standard Deviation calculation
    std_dev_val = df_clean[col].std()
    print(f"Standard Deviation : {std_dev_val:.2f}")
```

Skewness and Kurtosis

```python
print("\n==================================================")
print("3. SKEWNESS AND KURTOSIS")
print("==================================================")

for col in columns_to_analyze:
    print(f"\n--- Feature: {col} ---")

    # 1. Skewness calculation
    # > 0 means right-skewed; < 0 means left-skewed
    skew_val = df_clean[col].skew()
    print(f"Skewness         : {skew_val:.4f}")

    # 2. Kurtosis calculation (Pandas computes Excess Kurtosis)
    # > 0 means sharp peak/heavy tails; < 0 means flat peak/thin tails
    kurt_val = df_clean[col].kurtosis()
    print(f"Excess Kurtosis  : {kurt_val:.4f}")
print("\n==================================================")
```

```python
# Advanced Grouped Analysis (Malnutrition Rates Across Water Infrastructure Tiers)
# Create custom utility coverage categorizations
df['water_coverage_pct'] = np.where(df['anganwadi_centers'] > 0, (df['awc_with_drinking_water'] / df['anganwadi_centers']) * 100, 0)
df['water_infra_tier'] = pd.cut(df['water_coverage_pct'], bins=[-1, 33, 66, 100], labels=['Poor (<33%)', 'Moderate (33-66%)', 'Good (>66%)'])

# Generate and save Grouped Barplot
plt.clf()
pivot_data = df.groupby('water_infra_tier', observed=False)[['pct_stunted_children', 'pct_underweight_children']].mean().reset_index()
pivot_melted = pd.melt(pivot_data, id_vars=['water_infra_tier'], value_vars=['pct_stunted_children', 'pct_underweight_children'],
                       var_name='Malnutrition Metric', value_name='Percentage')
sns.barplot(data=pivot_melted, x='water_infra_tier', y='Percentage', hue='Malnutrition Metric', palette='muted')
plt.title('Average Malnutrition Rates by Drinking Water Infrastructure Tier')
plt.xlabel('Drinking Water Availability Tier')
plt.ylabel('Percentage (%)')
plt.tight_layout()
plt.savefig('multivariate_grouped_barplot.png')
```

INTERPRETATION:

The Business Story: This chart provides an insight into the broader ecology of health: districts classified under the "Poor" drinking water tier (<33% center coverage) exhibit higher average rates of both stunting and underweight children compared to districts with "Good" water infrastructure. Clean water prevents recurrent waterborne infections like diarrhea, which directly hinder a child's ability to absorb vital nutrients.

Operational Action: This highlights that nutritional outcomes cannot be solved by food delivery alone. The Ministry should form joint task forces with water and sanitation departments (such as the Jal Jeevan Mission) to prioritize installing clean water pipelines in districts with low infrastructure coverage. This dual approach helps maximize the impact of supplemental food programs.


**Summary of Findings**

The analysis of the National Nutrition Mission (POSHAN Abhiyaan) monitoring metrics reveals a critical story about the health and structural integrity of child care centers (Anganwadi Centers - AWCs) across India.

While administrative logging and reach are exceptionally vast—with massive resource hubs concentrated across high-population states like Uttar Pradesh, Madhya Pradesh, and Bihar—regional service delivery is highly uneven. On average, nearly 1 in 3 children (31.78%) under the age of six suffer from chronic skeletal growth delays (stunting). Furthermore, while sudden weight loss crisis metrics (wasting) remain low on a national level, specific "hotspot" districts experience severe spikes that require localized emergency support.

Crucially, the data reveals that nutrition cannot be treated purely as a food delivery problem. Structural issues at the centers—specifically the absence of clean drinking water and functional sanitation—show a clear relationship with higher district-level malnutrition benchmarks. This highlights the urgent need to transition from generic welfare distribution to an integrated, data-driven healthcare model.

**Key Insights**

**The Infrastructure-Nutrition Connection:** Districts with poor
infrastructure (where fewer than 33% of Anganwadi centers have working drinking water) show significantly higher average rates of both stunting and underweight children. Safe water is essential to prevent chronic infections that block nutrient absorption.

**The Chronic Stunting Plateau:** Severe-to-moderate stunting follows a balanced, widespread distribution centered around 31.78%. This indicates that chronic malnutrition is a structural, nationwide challenge rather than a collection of isolated regional issues.

**Acute Wasting is Highly Localized:** Child wasting has a low nationwide median ($\sim 4\%$) but displays extreme right-tailed outliers (kurtosis of $+9.70$). This means acute malnutrition occurs in intense, localized pockets that require targeted emergency response rather than uniform nationwide spending.

**The "Underweight" Metric Simplifies Screening:** A strong, linear correlation exists between children categorized as underweight and those categorized as stunted. This makes the underweight metric a highly reliable and cost-effective first-line screening tool for field workers.

**Operational Overcapacity Bottlenecks:** Analysis of beneficiary distribution shows that several districts have a low number of centers managing disproportionately massive beneficiary pools. This heavily strains local Anganwadi workers and compromises care quality.



**Types of Data Analysis Performed**

**Descriptive Analysis** (What is happening?)
Methodology: Evaluated through basic data parsing, measures of central tendency, and univariate distributions.

Finding: The average district contains approximately 1,829 centers tracking roughly 127,530 eligible beneficiaries. The baseline national averages for stunting and underweight children sit at 31.78% and 14.19%, respectively.

**Diagnostic Analysis** (Why is it happening?)
Methodology: Conducted via multi-variable correlation matrix mapping, outlier boxplot testing, and bivariate barplot distributions.

Finding: Bivariate analysis proved that structural growth failures do not happen in a vacuum. Segmenting the data revealed that districts in the lowest tier of clean water availability consistently show higher average stunting and underweight rates, diagnosing a clear link between poor sanitation and growth failure.

**Predictive Analysis **(What could happen next?)Methodology: Identified by evaluating distribution shapes, skewness coefficients, and growth capacity curves.

Finding: The positive skewness ($+2.35$) and high kurtosis ($+9.60$) in beneficiary volumes indicate that without immediate structural intervention, a small percentage of mega-districts will completely overwhelm local center capacities as populations grow, causing service quality to drop.

**Prescriptive Analysis** (How should we react?)
Methodology: Developed by building tailored Key Performance Indicators (KPIs), such as the Infrastructure Utility Percentage, and tracking absolute vulnerable populations.

Finding: Transforming raw percentages into absolute numbers allows logistics teams to determine exactly how many metric tons of therapeutic food supplies and how many sanitation installations are required per district.



**Recommendations & Decision Support**

**Implement a "Dual-Trigger" Resource Strategy:** For Chronic Malnutrition (Stunting): Allocate long-term, steady financial budgets to the top 10 highest-burden states identified in the bar chart analysis for food fortification programs.

**For Acute Malnutrition (Wasting):** Deploy short-term emergency funds and Ready-to-Use Therapeutic Food (RUTF) packets strictly to the right-tail outlier districts flagged in the boxplot analysis.

**Form an Infrastructure Cross-Task Force:** The Ministry of Women and Child Development should partner directly with sanitation programs (like India's Jal Jeevan Mission). Funding should be prioritized to install clean water systems in the "Poor Infrastructure Tier" districts to maximize the health impact of supplemental food programs.

**Relieve Operational Overcapacity:** Use the center-to-beneficiary scatterplot to identify districts that sit far above the normal density line. In these areas, divide large, over-extended centers into smaller, localized sub-units to prevent staff burnout and improve care.
